```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(ascending=False)

region_sales
```

```python
region_sales.plot(kind="bar", figsize=(8,5))

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.savefig("Images/sales_by_Region.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

category_sales
```

```python
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.savefig("Images/sales_by_category.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
monthly_sales = df.groupby("Month")["Sales"].sum()

monthly_sales
```

```python
monthly_sales.plot(kind="line", marker="o", figsize=(10,5))

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.savefig("Images/Monthly_Sales_Trend.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
df["Month_Number"] = df["Order Date"].dt.month
```

```python
monthly_sales = df.groupby(["Month_Number","Month"])["Sales"].sum().reset_index()
```

```python
monthly_sales = monthly_sales.sort_values("Month_Number")
```

```python
monthly_sales
```

```python
plt.figure(figsize=(10,5))

plt.plot(
    monthly_sales["Month"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.grid(True)

plt.savefig("Images/Monthly_Sales_Trend1.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
top_products = (
    df.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_products
```

```python
plt.figure(figsize=(12,6))

top_products.plot(kind="bar")

plt.title("Top 10 Products by Sales")

plt.xlabel("Product Name")

plt.ylabel("Sales")

plt.xticks(rotation=75)

plt.savefig("Images/Top_10_Products_by_Sales.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
subcategory_profit = (
    df.groupby("Sub-Category")["Profit"]
    .sum()
    .sort_values()
)

subcategory_profit
```

```python
plt.figure(figsize=(12,6))

subcategory_profit.plot(kind="barh")

plt.title("Profit by Sub-Category")

plt.xlabel("Profit")

plt.savefig("Images/Profit_by_Sub_Category.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
top_customers = (
    df.groupby("Customer Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

top_customers
```

```python
plt.figure(figsize=(12,6))

top_customers.plot(kind="bar", color="green")

plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer")
plt.ylabel("Sales")

plt.xticks(rotation=75)

plt.savefig("Images/Top_10_Customers_by_Sales.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
ship_mode = df.groupby("Ship Mode")["Sales"].sum()

ship_mode
```

```python
plt.figure(figsize=(8,5))

ship_mode.plot(kind="pie", autopct="%1.1f%%")

plt.ylabel("")

plt.title("Sales by Ship Mode")

plt.savefig("Images/Sales_by_Ship_Mode.png", dpi=300, bbox_inches="tight")

plt.show()
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum()

segment_sales
```

```python
plt.figure(figsize=(8,5))

segment_sales.plot(kind="bar", color="orange")

plt.title("Sales by Customer Segment")

plt.xlabel("Segment")

plt.ylabel("Sales")

plt.savefig("Images/Sales_by_Customer_Segment.png", dpi=300, bbox_inches="tight")

plt.show()
```

# Conclusion

This analysis examined Superstore sales data to identify key business insights.

### Key Findings

- Technology generated the highest sales among all categories.
- The West region recorded the highest sales.
- November achieved the highest monthly sales.
- Copiers were the most profitable sub-category.
- Tables and Bookcases generated losses and require business attention.
- The Canon imageCLASS 2200 Advanced Copier was the highest-selling product.

### Business Recommendations

- Increase marketing campaigns during November and December.
- Focus on promoting Technology products.
- Review the pricing and cost strategy for Tables and Bookcases.
- Maintain inventory for top-selling products to avoid stock shortages.
